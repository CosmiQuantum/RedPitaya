```python
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy.signal import welch
```

```python
sampling_rate_wanted = 32000
clock = 125 * (10**6)
decimation = round(clock / sampling_rate_wanted)
sampling_rate_real = clock / decimation

# 1) pr off
file = Path(r"data\PRoff.csv")
data = pd.read_csv(file)

samples = len(data)
voltages = data["Voltage (V)"].to_numpy()
f_datad1, data_psdd1 = welch(voltages, fs = sampling_rate_real, nperseg = samples // 8)

# 2) pr on
file = Path(r"data\PRon.csv")
data = pd.read_csv(file)

samples = len(data)
voltages = data["Voltage (V)"].to_numpy()
f_datad2, data_psdd2 = welch(voltages, fs = sampling_rate_real, nperseg = samples // 8)

# 3) direct to laser
file = Path(r"data\LaserDirect.csv")
data = pd.read_csv(file)

samples = len(data)
voltages = data["Voltage (V)"].to_numpy()
f_datad3, data_psdd3 = welch(voltages, fs = sampling_rate_real, nperseg = samples // 8)

# 4) all noise
file = Path(r"data\AllNoise.csv")
data = pd.read_csv(file)

samples = len(data)
voltages = data["Voltage (V)"].to_numpy()
f_datad4, data_psdd4 = welch(voltages, fs = sampling_rate_real, nperseg = samples // 8)

# 5) off res
file = Path(r"data\OffRes.csv")
data = pd.read_csv(file)

samples = len(data)
voltages = data["Voltage (V)"].to_numpy()
f_datad5, data_psdd5 = welch(voltages, fs = sampling_rate_real, nperseg = samples // 8)

# 7) no device laser off
file = Path(r"data\NoDeviceLASEROFF.csv")
data = pd.read_csv(file)

samples = len(data)
voltages = data["Voltage (V)"].to_numpy()
f_datad7, data_psdd7 = welch(voltages, fs = sampling_rate_real, nperseg = samples // 8)

# 8) no device laser on
file = Path(r"data\NoDeviceLASERON.csv")
data = pd.read_csv(file)

samples = len(data)
voltages = data["Voltage (V)"].to_numpy()
f_datad8, data_psdd8 = welch(voltages, fs = sampling_rate_real, nperseg = samples // 8)

plt.loglog(f_datad1, data_psdd1, label = "PR off", lw = 0.6)
plt.loglog(f_datad2, data_psdd2, label = "PR on", lw = 0.6)
plt.loglog(f_datad3, data_psdd3, label = "Direct to laser connection", lw = 0.6)
plt.loglog(f_datad4, data_psdd4, label = "Full system noise", lw = 0.6)
plt.loglog(f_datad5, data_psdd5, label = "Off res noise", lw = 0.6)
plt.loglog(f_datad7, data_psdd7, label = "No device laser off", lw = 0.6)
plt.loglog(f_datad8, data_psdd8, label = "No device laser on", lw = 0.6)
plt.xlabel("Frequency (Hz)")
plt.ylabel("Voltage (V^2 / Hz)")
plt.legend()
plt.gcf().set_size_inches(9,7)
plt.title("Noise PSDs")
```
